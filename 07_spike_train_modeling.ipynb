# Spike Train Modeling of Retinal Neural Activity

## Objective

This notebook simulates simplified neural spike train behavior associated with retinal information encoding.

## Methods

The notebook uses:
- NumPy for spike generation
- probabilistic firing models
- Matplotlib for spike train visualization

## Biological Context

Retinal ganglion cells communicate visual information through sequences of action potentials. Spike train modeling helps represent how neural activity may encode time-varying sensory information.

## Expected Outcome

The simulation is expected to generate discrete spike events that approximate simplified retinal neural firing behavior over time.

```python
import numpy as np
import matplotlib.pyplot as plt

time = np.linspace(0, 10, 1000)

firing_rate = 20
spike_probability = firing_rate / len(time)

spikes = np.random.rand(len(time)) < spike_probability

plt.figure(figsize=(10,4))
plt.eventplot(time[spikes], lineoffsets=1, linelengths=0.8)

plt.title("Simulated Retinal Ganglion Cell Spike Train")
plt.xlabel("Time")
plt.ylabel("Spike Events")

plt.savefig("images/spike_train.png", dpi=300, bbox_inches="tight")
plt.show()
```
