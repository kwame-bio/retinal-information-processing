# Signal Filtering in Retinal Information Processing

## Objective

This notebook investigates computational filtering approaches relevant to biological signal processing.

## Methods

The notebook uses:
- NumPy for noisy signal generation
- moving-average or convolution-based filtering
- Matplotlib for before-and-after signal comparison

## Biological Context

Biological sensory systems must extract meaningful information from noisy input. Signal filtering models help explore how retinal systems may preserve important patterns while reducing irrelevant variability.

## Expected Outcome

The notebook is expected to show how filtering can recover smoother signal structure from noisy retinal-like activity.

```python
import numpy as np
import matplotlib.pyplot as plt

time = np.linspace(0, 10, 500)

clean_signal = np.sin(time)
noise = np.random.normal(0, 0.3, len(time))
noisy_signal = clean_signal + noise

window_size = 15
filtered_signal = np.convolve(
    noisy_signal,
    np.ones(window_size) / window_size,
    mode="same"
)

plt.figure(figsize=(10,6))

plt.plot(time, noisy_signal, label="Noisy Signal", alpha=0.5)
plt.plot(time, filtered_signal, label="Filtered Signal")
plt.plot(time, clean_signal, label="Original Clean Signal", linestyle="--")

plt.title("Filtering Noisy Retinal Signals")
plt.xlabel("Time")
plt.ylabel("Signal Intensity")
plt.legend()

plt.show()
```
