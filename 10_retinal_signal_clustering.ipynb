# Clustering Analysis of Retinal Signal Activity

## Objective

This notebook applies unsupervised machine learning methods to identify structure in retinal-like neural activity.

## Methods

The notebook uses:
- NumPy for simulated feature generation
- scikit-learn for clustering analysis
- Matplotlib for cluster visualization

## Biological Context

Neural populations may contain structured subgroups based on signal behavior, response properties, or biological function. Clustering helps reveal organization that may not be obvious from raw data.

## Expected Outcome

The notebook is expected to identify distinct signal groups within simulated retinal activity patterns.

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# Simulated retinal signal clusters

X, y = make_blobs(
    n_samples=300,
    centers=3,
    cluster_std=1.2,
    random_state=42
)

# KMeans clustering

kmeans = KMeans(n_clusters=3)

kmeans.fit(X)

predicted_clusters = kmeans.predict(X)

# Plot results

plt.figure(figsize=(8,6))

plt.scatter(
    X[:,0],
    X[:,1],
    c=predicted_clusters
)

plt.title("Clustering Simulated Retinal Signal Patterns")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.savefig("images/clustering.png", dpi=300, bbox_inches="tight")
plt.show()
```
