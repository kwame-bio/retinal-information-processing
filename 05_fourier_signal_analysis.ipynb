# Fourier Analysis of Retinal Signal Frequencies

## Objective

This notebook explores frequency-domain decomposition of simulated retinal waveform activity.

## Methods

The notebook uses:
- NumPy for signal simulation
- Fourier Transform methods for frequency analysis
- Matplotlib for frequency-spectrum visualization

## Biological Context

Neural signals often contain temporal and oscillatory structure. Fourier analysis helps identify frequency components that may represent rhythmic or patterned activity in retinal signal processing.

## Expected Outcome

The notebook is expected to reveal dominant frequency components within simulated retinal signals.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate simulated retinal signal

time = np.linspace(0, 10, 500)

signal = (
    np.sin(2 * np.pi * 1 * time)
    + 0.5 * np.sin(2 * np.pi * 3 * time)
)
```

```python
# Fourier Transform

fourier = np.fft.fft(signal)
frequencies = np.fft.fftfreq(len(signal), d=(time[1]-time[0]))

magnitude = np.abs(fourier)
```

```python
# Plot frequency spectrum

plt.figure(figsize=(10,5))

plt.plot(frequencies[:250], magnitude[:250])

plt.title("Frequency Spectrum of Simulated Retinal Signal")
plt.xlabel("Frequency")
plt.ylabel("Magnitude")

plt.savefig("images/fourier.png", dpi=300, bbox_inches="tight")
plt.show()
```
