# Information Theory and Retinal Encoding

## Objective

This notebook explores how information-theoretic concepts can be used to model retinal signal encoding under biological noise.

## Methods

The notebook uses:
- NumPy for signal simulation
- probability distributions for entropy estimation
- Matplotlib for visualization
- simple Shannon entropy calculations

## Biological Context

The retina must transmit useful visual information despite noise and biological variability. Information theory provides tools for studying uncertainty, signal structure, and encoding efficiency in neural systems.

## Expected Outcome

The simulation is expected to show how biological noise changes signal uncertainty and how entropy can describe retinal-like encoding efficiency.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulated retinal signal
np.random.seed(42)
time = np.linspace(0, 10, 500)

clean_signal = np.sin(2 * np.pi * 1 * time)
noise = np.random.normal(0, 0.5, size=len(time))
noisy_signal = clean_signal + noise

# Convert signals into probability distributions
clean_counts, _ = np.histogram(clean_signal, bins=30, density=True)
noisy_counts, _ = np.histogram(noisy_signal, bins=30, density=True)

clean_probs = clean_counts / np.sum(clean_counts)
noisy_probs = noisy_counts / np.sum(noisy_counts)

clean_probs = clean_probs[clean_probs > 0]
noisy_probs = noisy_probs[noisy_probs > 0]

# Shannon entropy
clean_entropy = -np.sum(clean_probs * np.log2(clean_probs))
noisy_entropy = -np.sum(noisy_probs * np.log2(noisy_probs))

print("Clean Signal Entropy:", round(clean_entropy, 3))
print("Noisy Signal Entropy:", round(noisy_entropy, 3))

plt.figure(figsize=(10, 5))
plt.plot(time, clean_signal, label="Clean Retinal Signal")
plt.plot(time, noisy_signal, label="Noisy Retinal Signal", alpha=0.7)

plt.title("Retinal Signal Encoding Under Biological Noise")
plt.xlabel("Time")
plt.ylabel("Signal Intensity")
plt.legend()
plt.show()
```
