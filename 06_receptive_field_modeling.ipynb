# Computational Modeling of Retinal Receptive Fields

## Objective

This notebook models simplified receptive field structures involved in early-stage visual processing.

## Methods

The notebook uses:
- NumPy for Gaussian function generation
- Matplotlib for response visualization
- center-surround modeling to simulate retinal receptive field behavior

## Biological Context

Retinal ganglion cells often respond through center-surround receptive field organization. This structure helps enhance contrast, detect edges, and transform visual input before transmission to the brain.

## Expected Outcome

The notebook is expected to show how excitatory center and inhibitory surround responses combine to shape visual signal processing.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create simulated visual field

x = np.linspace(-10, 10, 400)

center_signal = np.exp(-(x**2) / 4)

surround_signal = 0.6 * np.exp(-(x**2) / 20)
```

```python
# Center-surround receptive field

receptive_field = center_signal - surround_signal
```

```python
plt.figure(figsize=(10,6))

plt.plot(x, center_signal, label="Center Excitation")
plt.plot(x, surround_signal, label="Surround Inhibition")
plt.plot(x, receptive_field, label="Combined Receptive Field")

plt.title("Simulated Retinal Receptive Field")
plt.xlabel("Visual Space")
plt.ylabel("Response Strength")

plt.legend()

plt.show()
```
