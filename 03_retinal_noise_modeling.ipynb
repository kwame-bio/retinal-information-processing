# Biological Noise Modeling in Retinal Systems

## Objective

This notebook investigates how stochastic biological noise influences retinal signal fidelity and neural variability.

## Methods

The notebook uses:
- NumPy for signal generation
- random noise simulation
- Matplotlib for clean vs. noisy signal visualization

## Biological Context

Neural systems operate under biological variability caused by molecular, cellular, and electrical noise. Modeling noisy retinal signals helps explore how useful information can persist despite imperfect biological transmission.

## Expected Outcome

The simulation is expected to show how noise alters signal structure while preserving underlying patterns that may still be recoverable computationally.

```python
import numpy as np
import matplotlib.pyplot as plt

# Time axis
time = np.linspace(0, 10, 300)

# Original retinal signal
clean_signal = np.sin(time)

# Biological noise
noise = np.random.normal(0, 0.2, len(time))

# Noisy retinal signal
noisy_signal = clean_signal + noise

# Plotting
plt.figure(figsize=(10,6))

plt.plot(time, clean_signal, label="Clean Signal")
plt.plot(time, noisy_signal, label="Noisy Signal", alpha=0.7)

plt.title("Retinal Signal With Biological Noise")
plt.xlabel("Time")
plt.ylabel("Signal Intensity")
plt.legend()

plt.show()
```
