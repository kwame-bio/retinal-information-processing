```python
import numpy as np
import matplotlib.pyplot as plt

# Time axis
time = np.linspace(0, 10, 300)

# Original retinal signal
clean_signal = np.sin(time)

# Biological noise
noise = np.random.normal(0, 0.2, len(time))

# Noisy retinal signal
noisy_signal = clean_signal + noise

# Plotting
plt.figure(figsize=(10,6))

plt.plot(time, clean_signal, label="Clean Signal")
plt.plot(time, noisy_signal, label="Noisy Signal", alpha=0.7)

plt.title("Retinal Signal With Biological Noise")
plt.xlabel("Time")
plt.ylabel("Signal Intensity")
plt.legend()

plt.show()
```
