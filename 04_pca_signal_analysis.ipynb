```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Simulated retinal signals
time = np.linspace(0, 10, 300)

photoreceptor = np.sin(time)
bipolar = np.sin(time + 0.5)
ganglion = np.sin(time + 1.0)

# Combine signals
signals = pd.DataFrame({
    "Photoreceptor": photoreceptor,
    "Bipolar": bipolar,
    "Ganglion": ganglion
})

# PCA analysis
pca = PCA(n_components=2)
principal_components = pca.fit_transform(signals)

# Plot PCA
plt.figure(figsize=(8,6))

plt.scatter(
    principal_components[:,0],
    principal_components[:,1]
)

plt.title("PCA of Simulated Retinal Signals")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.show()
```
